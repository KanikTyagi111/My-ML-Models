#  Handling Output Labels (Target Encoding)

```python
unique_emotions = df['emotions'].unique()
```

```python
emotions_num = {}
```

```python
i = 0
for e in unique_emotions:
    emotions_num[e] = i
    i=+1
df['emotions'] = df['emotions'].map(emotions_num)
```

#  Lowercasing 

```python
df['text'] = df['text'].str.lower()   # df['text'] = df['text'].apply(lambda x : x.lower())
```

# Removing Punctuation

```python
import string

def remove_punch(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punch)
```

```python
df
```

# Removing Numbers from text

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

# Remove Emojies

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

df['text'] = df['text'].apply(remove_emojis)
```
